```python
import pandas as pd

def clean_data(df):
    # Filter rows based on columns: 'technology', 'scenario_geography'
    df = df[(df['technology'].str.contains("gascap", regex=False, na=False, case=False)) & (df['scenario_geography'].str.contains("R10MIDDLE_EAST", regex=False, na=False, case=False))]
    df = df.assign(npv_diff = df["latesudden_npv"] - df["baseline_npv"])
    # Performed 2 aggregations grouped on columns: 'sector', 'technology', 'scenario_geography'
    df = df.groupby(['sector', 'technology', 'scenario_geography']).agg(baseline_npv_sum=('baseline_npv', 'sum'), latesudden_npv_sum=('latesudden_npv', 'sum')).reset_index()
    df = df.assign(npv_diff = df["latesudden_npv_sum"] - df["baseline_npv_sum"])
    return df

# Loaded variable 'df' from URI: /Users/bertrandgallice/code/Theia-Finance-Labs/crispy-kedro/workspace/results_V8/REMIND-MAgPIE_DeepElec__asset_granularity_with_staggered_shock_and_retirement/asset_npv.csv
df = pd.read_csv(r'/Users/bertrandgallice/code/Theia-Finance-Labs/crispy-kedro/workspace/results_V8/REMIND-MAgPIE_DeepElec__asset_granularity_with_staggered_shock_and_retirement/asset_npv.csv')

df_clean1 = clean_data(df.copy())
df_clean1.head()
```
