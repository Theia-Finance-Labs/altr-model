```python
import pandas as pd

def clean_data(df):
    # Derive column 'scenario_name' from columns: 'scenario', 'scenario_provider'
    df.insert(2, 'scenario_name', df.apply(lambda row : "AR6_"+row["scenario_provider"]+"_"+row["scenario"], axis=1))
    # Filter rows based on column: 'scenario_name'
    df = df[(df['scenario_name'].str.contains("\"AR6_AIM/CGE 2.2_EN_NPi2020_1200f\"", regex=False, na=False, case=False)) | (df['scenario_name'].str.contains("AR6_AIM/CGE 2.2_EN_NPi2020_900f", regex=False, na=False, case=False)) | (df['scenario_name'].str.contains("AR6_COFFEE 1.1_CO_CurPol", regex=False, na=False, case=False)) | (df['scenario_name'].str.contains("AR6_COFFEE 1.1_EN_NPi2020_400f_lowBECCS", regex=False, na=False, case=False)) | (df['scenario_name'].str.contains("AR6_REMIND-MAgPIE 2.1-4.3_DeepElec_SSP2_Base", regex=False, na=False, case=False)) | (df['scenario_name'].str.contains("AR6_REMIND-MAgPIE 2.1-4.3_DeepElec_SSP2_ HighRE_Budg900", regex=False, na=False, case=False)) | (df['scenario_name'].str.contains("AR6_MESSAGEix-GLOBIOM_1.0_CO_CurPol", regex=False, na=False, case=False)) | (df['scenario_name'].str.contains("AR6_MESSAGEix-GLOBIOM_GEI 1.0_SSP2_openres_lc_50", regex=False, na=False, case=False)) | (df['scenario_name'].str.contains("AR6_REMIND-MAgPIE 2.1-4.2_SusDev_SDP-NPi", regex=False, na=False, case=False)) | (df['scenario_name'].str.contains("AR6_REMIND-MAgPIE 2.1-4.2_SusDev_SDP-PkBudg1000", regex=False, na=False, case=False)) | (df['scenario_name'].str.contains("AR6_WITCH 5.0_CO_CurPol", regex=False, na=False, case=False)) | (df['scenario_name'].str.contains("AR6_WITCH 5.0_CO_Bridge", regex=False, na=False, case=False))]
    # Filter rows based on column: 'carbon_price_usd_per_tco2'
    df = df[df['carbon_price_usd_per_tco2'].notna()]
    # Grouped on column: 'scenario_name'
    df = df.groupby(['scenario_name', "technology"]).count().reset_index()[['scenario_name', "technology"]]
    return df

# Loaded variable 'df' from URI: /Users/bertrandgallice/code/Theia-Finance-Labs/crispy-kedro/data/05_model_input/AR6_scenarios_carbon_price_fixed.csv
df = pd.read_csv(r'/Users/bertrandgallice/code/Theia-Finance-Labs/crispy-kedro/data/05_model_input/AR6_scenarios_carbon_price_fixed.csv')

df_clean = clean_data(df.copy())
```
