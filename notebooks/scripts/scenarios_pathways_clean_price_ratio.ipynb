```python
import pandas as pd
```

```python
df = pd.read_csv("../data/07_model_output/scenarios_pathways_clean.csv")
```

```python
df["price_ratio"] = df["scenario_price_scaled"] / df["scenario_price"] 
```

```python
result = df.groupby(["scenario", "scenario_type","scenario_geography", "technology"])["price_ratio"].agg([
    ('mean', 'mean'),
    ('median', 'median'),
    ('min', 'min'),
    ('max', 'max'),
    ('percentile_2.5', lambda x: x.quantile(0.025)),
    ('percentile_5', lambda x: x.quantile(0.05)),
    ('percentile_95', lambda x: x.quantile(0.95)),
    ('percentile_97.5', lambda x: x.quantile(0.975))
]).reset_index()
result = result.sort_values(by=["technology", "scenario_geography", "scenario"], ascending=False)
result
```

```python
result.to_csv("../price_scaling_ratios.csv", index=False)
```
