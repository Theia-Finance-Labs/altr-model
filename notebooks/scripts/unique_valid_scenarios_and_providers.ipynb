```python
import pandas as pd

def clean_data(df):
    # Filter rows based on column: 'sector'
    df = df[df['sector'].str.contains("power", regex=False, na=False, case=False)]
    required_techs = {"OilCap", "GasCap", "CoalCap"}
    # Find providers that have ALL required technologies
    valid_providers = (
        df.groupby("scenario_provider")["technology"]
          .apply(lambda x: required_techs.issubset(set(x)))
    )
    # Keep only valid providers
    df = df[df["scenario_provider"].isin(valid_providers[valid_providers].index)]
    # Grouped on columns: 'scenario_provider', 'scenario'
    df = df.groupby(['scenario_provider', 'scenario']).count().reset_index()[['scenario_provider', 'scenario']]
    return df

# Loaded variable 'df' from URI: /Users/bertrandgallice/code/Theia-Finance-Labs/crispy-kedro/data/05_model_input/AR6_scenarios_carbon_price_fixed.csv
df = pd.read_csv(r'/Users/bertrandgallice/code/Theia-Finance-Labs/crispy-kedro/data/05_model_input/AR6_scenarios_carbon_price_fixed.csv')

df_clean = clean_data(df.copy())
df_clean.head()
```
